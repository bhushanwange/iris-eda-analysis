# Exploratory Data Analysis (EDA): Iris Dataset
An exploratory analysis of the Iris flower dataset covering summary statistics, feature distributions, pairwise relationships, and key analytical insights.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

# Load the Iris dataset
df = sns.load_dataset("iris")

print("Dataset Shape:", df.shape)
display(df.head())
df.info()
```

```python
# Generate descriptive statistics table
stats_table = df.describe().T
stats_table['missing_values'] = df.isnull().sum()
display(stats_table)

# Verify class balance
print("\nSpecies Distribution:")
print(df['species'].value_counts())
```

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for idx, col in enumerate(features):
    sns.histplot(data=df, x=col, kde=True, ax=axes[idx], color='steelblue', bins=15)
    axes[idx].set_title(f'Distribution of {col.replace("_", " ").title()}', fontsize=12)
    axes[idx].set_xlabel(col.replace("_", " ").title())
    axes[idx].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
pair_plot = sns.pairplot(df, hue="species", palette="Set2", diag_kind="hist", height=2.5)
pair_plot.fig.suptitle("Pairwise Scatter Plots by Iris Species", y=1.02, fontsize=14)
plt.show()
```

## Summary of Findings

1. **Clean & Balanced Dataset:** The Iris dataset has **150 rows** and **5 columns** with **0 missing values**, evenly split across three species (**50 samples each** for *setosa*, *versicolor*, and *virginica*).
2. **Feature Distributions:** 
   * `sepal_length` and `sepal_width` display roughly normal (bell-shaped) distributions.
   * `petal_length` and `petal_width` show clear **bimodal distributions**, highlighting a distinct gap between *Iris setosa* and the other two species.
3. **Species Separability:** 
   * ***Iris setosa*** is completely linearly separable from *versicolor* and *virginica*, characterized by much smaller petal lengths (< 2.0 cm) and petal widths (< 0.7 cm).
   * ***Iris versicolor*** and ***Iris virginica*** overlap slightly in sepal measurements, though *virginica* consistently has larger petal dimensions.
4. **Strong Feature Correlation:** `petal_length` and `petal_width` exhibit a strong positive linear correlation, making them the strongest predictors for distinguishing flower species.
